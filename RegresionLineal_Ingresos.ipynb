MODELO DE REGRESION LINEAL SIMPLE

```python
import pandas as pd

datos = pd.read_csv('ingreso.csv')

datos
```

REDONDEO DE VALORES EN CADA COLUMNA

```python
datos = datos[['ingreso', 'horas']].round(2)
datos
```

CORRELACION ENTRE AMBAS COLUMNAS

```python
datos[['ingreso', 'horas']].corr()
```

GRAFICO DE DISPERSION PARA VISUALIZAR LA CORRELACION LINEAL

```python
import matplotlib.pyplot as plt

plt.ylabel('Ingreso($)')
plt.xlabel('Promedio Hrs Semanales')
plt.scatter(datos['horas'], datos['ingreso'], color='skyblue')
plt.show()
```

CREACION DE MODELO DE REGRESION LINEAL SIMPLE

```python
from sklearn import linear_model

regresion = linear_model.LinearRegression()

horas = datos['horas'].values.reshape((-1, 1))

modelo = regresion.fit(horas, datos['ingreso'])
```

PENDIENTE & PTO DE INTESERCCION DE LA RECTA (y=mx+b)

```python
print('Interseccion (b) =', modelo.intercept_)
print('Pendiente (m) =', modelo.coef_)
```

VALORES A PREDECIR EN EL MODELO

```python
x = [[39], [40], [43], [44]]

y = modelo.predict(x).round(2)

print(y)
```

GRAFICO DE DISPERSION PARA GRAFICAR REGRESION LINEAL DEL MODELO 

```python
plt.scatter(x, y , color='red')
plt.plot(x, y, color='yellow')

plt.ylabel('Ingreso($)')
plt.xlabel('Promedio Hrs Semanales')
plt.scatter(datos['horas'], datos['ingreso'], color='skyblue')
plt.show()
```
